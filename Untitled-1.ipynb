```python
import sqlite3
import folium
import pandas as pd
pd.set_option("display.max_columns", None)
```

```python
cnx = sqlite3.connect('data.db')

# tables = pd.read_sql_query(
#     "SELECT name FROM sqlite_master WHERE type='table';",
#     cnx
# )

# print(tables)

# df = pd.read_sql_query("SELECT num, localidad, direcc, lat, lng FROM prices group by num, localidad, direcc, lat, lng", cnx)
df = pd.read_sql_query("SELECT * FROM prices", cnx)
df["fecha"] = pd.to_datetime(
    df["fecha"],
    format="%d-%m-%Y %H:%M"
)

df = df.sort_values("gasolina_95", ascending=True)
df = df.sort_values("fecha", ascending=False)


# display(df)
df[df["num"] == 340013538]
```

```python
cnx.execute("DELETE FROM prices WHERE num = ?", (340013538,))
cnx.commit()

cnx.close()
```

```python
# Center the map on the average location in your dataframe
m = folium.Map(
    location=[df["lat"].mean(), df["lng"].mean()],
    zoom_start=10,
    # width='60%',
    # height='60%'
)

# Add one marker for every row
for _, row in df.iterrows():
    folium.Marker(
        location=[row["lat"], row["lng"]]
    ).add_to(m)

m
```
